```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python
df = pd.read_csv("Athletes_winter_games.csv")
df
```

# duplicate dataset

```python
df1 = df.copy()
df1
```

```python
# creating the target variable medal won
```

```python
df1['Medal_Won'] = df1['Medal'].apply(lambda x: 0 if x=="No Medal" else 1)
df1
```

# droping the first column name

```python
df1 = df1.drop(columns=['Unnamed: 0'])
df1
```

# imformation of dataset

```python
df1.info()
```

# replacing Nan values 

```python
df1['Medal'] = df1 ['Medal'].fillna('No Medal')
df1
```

```python
df1.head()
```

# fill missing age with median

```python
df1['Age'] = df1['Age'].fillna(df['Age'].median())
df1
```

# fill mising by sports median

```python
df1['Age'] = df1.groupby('Sport')['Age'].transform(lambda x: x.fillna(x.median()))
df1
```

# removing of some columns that has too many values

```python
features = ['Sex', 'Age', 'Team', 'NOC', 'Sport', 'Year']
df1_ml = df1[features + ['Medal_Won']]
df1
```

# Encode categorical columns

```python
df1_encoded = pd.get_dummies(df1_ml, drop_first=True)
df1
```

# train test split

```python
from sklearn.model_selection import train_test_split
```

```python
x = df1_encoded.drop('Medal_Won',axis=1)
y = df1_encoded['Medal_Won']
df1
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=42)
x_train
```

# age distribution

```python
plt.figure(figsize=(8,5))
plt.hist(df['Age'].dropna(), bins=20)
plt.xlabel('Age')
plt.ylabel('Count')
plt.title('Age Distribution of Winter Olympic Athletes')
plt.savefig('Age distribution winter olympic')
plt.show()
```

```python
# This plot shows how athlete ages are spread across the Winter Olympics.
# Most competitors are in their early to mid-20s, reflecting peak performance age for winter sports.
# It helps reveal general age trends among participants
```

# numbers of athletes by sports

```python
plt.figure(figsize=(12,6))
df['Sport'].value_counts().plot(kind='bar')
plt.title('Number of Athletes by Sport')
plt.xlabel('Sport')
plt.ylabel('Athlete Count')
plt.xticks(rotation=90)
plt.savefig("Numbers of Athletes by sports")
plt.show()
```

```python
# This chart.ws how many athletes participated in each Winter Olympic sport.
# It highlights the most popular sports with the highest athlete counts and gives a quick
# overview of participation across all categories
```

```python
plt.figure(figsize=(6,6))
df['Sex'].value_counts().plot(kind='pie', autopct='%1.1f%%')
plt.title('Gender Distribution of Athletes')
plt.ylabel('')
plt.savefig("Gender distribution")
plt.show()
```

```python
# This chart compares the number of male and female athletes in the Winter Olympics,
# showing the overall balance of participation between both genders.
```

# numbers of event per year

```python
events_per_year = df.groupby('Year')['Event'].count()
plt.figure(figsize=(10,5))
plt.plot(events_per_year.index, events_per_year.values)
plt.title('Events Count Across Winter Olympic Years')
plt.xlabel('Year')
plt.ylabel('Number of Events')
plt.grid(True)
plt.savefig("Events Count Across Winter Olympic Year")
plt.show()
```

```python
# This chart compares the number of male and female athletes in the Winter Olympics,
# showing the overall balance of participation between both genders.
```
